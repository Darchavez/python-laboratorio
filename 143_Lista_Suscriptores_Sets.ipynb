# 🧪 143 — Lista de Suscriptores con Sets

**Aprendiendo:** sets (conjuntos) para evitar duplicados, agregar con `add()`, eliminar con `discard()`, revisar si un elemento existe con `in` y contar con `len()`.

## Enunciado
Crear un programa para administrar una lista de suscriptores utilizando su email.

- Una persona se suscribe al boletín informativo utilizando su email.
- A medida que la lista crece, hay que asegurarse de que no haya suscriptores duplicados.
- También se debe poder agregar y eliminar suscriptores.

## 1. Crear la lista de suscriptores

```python
print('*** Lista de Suscriptores ***')

suscriptores = {'ana@mail.com', 'luis@mail.com', 'ana@mail.com', 'maria@mail.com'}

print(f'Suscriptores: {suscriptores}')
print(f'Total de suscriptores: {len(suscriptores)}')
```

## 2. Agregar un suscriptor

```python
nuevo_email = input('\nEmail para suscribirse: ').strip().lower()

if nuevo_email in suscriptores:
    print(f'{nuevo_email} ya está suscrito.')
else:
    suscriptores.add(nuevo_email)
    print(f'{nuevo_email} fue agregado a la lista.')

print(f'Total de suscriptores: {len(suscriptores)}')
```

## 3. Intentar agregar un email duplicado

```python
email_repetido = input('\nEmail para suscribirse: ').strip().lower()

if email_repetido in suscriptores:
    print(f'{email_repetido} ya está suscrito.')
else:
    suscriptores.add(email_repetido)
    print(f'{email_repetido} fue agregado a la lista.')

print(f'Total de suscriptores: {len(suscriptores)}')
```

## 4. Eliminar un suscriptor

```python
email_eliminar = input('\nEmail para darse de baja: ').strip().lower()

if email_eliminar in suscriptores:
    suscriptores.discard(email_eliminar)
    print(f'{email_eliminar} fue eliminado de la lista.')
else:
    print(f'{email_eliminar} no está en la lista.')

print(f'\nLista final: {sorted(suscriptores)}')
print(f'Total de suscriptores: {len(suscriptores)}')
```
