# 🧪 184 — Calculadora con Funciones

**Aprendiendo:** funciones que **reciben parámetros** y **devuelven valores** con `return` (incluso dos valores a la vez), una variable bandera (`salir`) para controlar un ciclo `while`, y comparaciones encadenadas (`1 <= opcion <= 4`).

## Enunciado
Crear una calculadora que muestre un menú con estas opciones:

| Opción | Operación |
|---|---|
| 1 | Suma |
| 2 | Resta |
| 3 | Multiplicación |
| 4 | División |
| 5 | Salir |

Para las opciones 1 a 4, se deben pedir dos valores y mostrar el resultado. El programa se repite hasta que el usuario elija **Salir**. Cada parte del programa debe estar en su propia función.

## 1. Función para mostrar el menú
Muestra las opciones y **devuelve** con `return` la opción que escogió el usuario.

```python
print('*** Calculadora con Funciones ***')

def mostrar_menu():
    print('''Operaciones que puedes realizar:
1. Suma
2. Resta
3. Multiplicación
4. División
5. Salir''')
    return int(input('Escoge una opción: '))
```

## 2. Función para pedir los valores
Una función puede devolver **dos valores a la vez**, separados por coma. Quien la llama los recibe en dos variables: `operando1, operando2 = pedir_valores()`.

```python
def pedir_valores():
    operando1 = float(input('Dame el valor 1: '))
    operando2 = float(input('Dame el valor 2: '))
    return operando1, operando2
```

## 3. Función para ejecutar la operación
Solo pide los valores si la opción está **entre 1 y 4**. Devuelve `salir`: será `True` únicamente cuando el usuario escoja la opción 5.

```python
def ejecutar_operacion(opcion, salir):
    # Solicitar los valores de los operandos
    if 1 <= opcion <= 4:
        operando1, operando2 = pedir_valores()
    resultado = 0
    if opcion == 1:    # Sumar
        resultado = operando1 + operando2
        print(f'El resultado de la suma es: {resultado}\n')
    elif opcion == 2:  # Restar
        resultado = operando1 - operando2
        print(f'El resultado de la resta es: {resultado}\n')
    elif opcion == 3:  # Multiplicación
        resultado = operando1 * operando2
        print(f'El resultado de la multiplicación es: {resultado}\n')
    elif opcion == 4:  # División
        resultado = operando1 / operando2
        print(f'El resultado de la división es: {resultado}\n')
    elif opcion == 5:  # Salir
        print('Saliendo del programa de calculadora, ¡hasta pronto!')
        salir = True
    else:
        print('Opción inválida, selecciona otra opción...\n')
    return salir
```

## 4. Programa principal
`salir` empieza en `False`. El ciclo `while not salir` se repite **mientras no** haya que salir, y en cada vuelta se actualiza con lo que devuelve `ejecutar_operacion()`.

```python
if __name__ == '__main__':
    salir = False
    while not salir:
        opcion = mostrar_menu()
        salir = ejecutar_operacion(opcion, salir)
```
