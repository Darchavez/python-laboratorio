# 🧪 174 — Potencia de un Número con Recursividad

**Aprendiendo:** funciones **recursivas** (una función que se llama a sí misma), la diferencia entre **caso base** y **caso recursivo**, y `return` para devolver el resultado.

## Enunciado
Crear una función recursiva que calcule la potencia de un número, recibiendo la **base** y el **exponente**.

Probar la función con estos casos:

| Base | Exponente | Resultado esperado |
|---|---|---|
| 2 | 3 | 8 |
| 5 | 0 | 1 |
| 4 | 5 | 1024 |

## 1. Definir la función recursiva
- **Caso base:** cuando el exponente llega a 0, devuelve 1 (todo número elevado a 0 es 1). Es la condición que **detiene** la recursión.
- **Caso recursivo:** multiplica la base por la potencia con el exponente **uno menos**. Cada llamada se acerca un paso al caso base.

```python
print('*** Potencia número usando funciones recursivas ***')

def potencia(base, exponente):
    # Caso base
    if exponente == 0:
        return 1
    else:  # Caso recursivo
        return base * potencia(base, exponente - 1)
```

## 2. Probar la función
Así se resuelve `potencia(2, 3)` por dentro:

`potencia(2, 3)` → `2 * potencia(2, 2)` → `2 * 2 * potencia(2, 1)` → `2 * 2 * 2 * potencia(2, 0)` → `2 * 2 * 2 * 1` = **8**

```python
print(f'2 elevado a la 3: {potencia(2, 3)}')
print(f'5 elevado a la 0: {potencia(5, 0)}')
print(f'4 elevado a la 5: {potencia(4, 5)}')
```
