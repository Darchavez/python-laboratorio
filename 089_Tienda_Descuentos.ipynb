# 🧪 089 — Tienda en línea con descuentos

**Aprendiendo:** condicionales (`if/elif/else`), operadores de comparación (`>=`) y lógicos (`and`), constantes, conversión de datos con `float(input())` y cálculo de porcentajes.

## Enunciado
Crear un sistema que ofrezca descuentos según el monto de la compra y si el cliente es miembro de la tienda.

El sistema debe solicitar:
- Monto de la compra
- Si es miembro de la tienda (Sí/No)

Se deben revisar las siguientes condiciones:

| Condición | Descuento |
|---|---|
| Compró 1,000 o más **y** es miembro | 10% |
| Es miembro (compra menor a 1,000) | 5% |
| Compró 1,000 o más (no es miembro) | 3% |
| No es miembro y compró menos de 1,000 | 0% |

Al final, mostrar el monto de la compra, el monto del descuento y el monto final a pagar.

## 1. Declaración de variables, constantes y entrada de datos

```python
print('*** Sistema Tienda en Línea con Descuentos ***')

# Condiciones
MONTO_COMPRA_DESC = 1000

monto_compra = float(input('¿Cual fue el monto de tu compra? '))
es_miembro = input('¿Eres miembro de la tienda (Sí/No)? ').strip().lower() in ('si', 'sí')

descuento = 0
```

## 2. Verificar cada caso

```python
# Verificar cada caso, con los datos proporcionados
if monto_compra >= MONTO_COMPRA_DESC and es_miembro:
    descuento = 0.1  # Descuento del 10%
elif es_miembro:
    descuento = .05  # Descuento del 5%
elif monto_compra >= MONTO_COMPRA_DESC:
    descuento = .03  # Descuento del 3%
else:
    descuento = 0
```

## 3. Hacemos los cálculos respectivos para obtener el monto final

```python
# Hacemos los cálculos respectivos para obtener el monto final
if descuento != 0:
    monto_descuento = monto_compra * descuento
    monto_final = monto_compra - monto_descuento
    print(f'\nFelicidades, has obtenido un descuento del {descuento * 100:.0f}%')
    print(f'Monto de la compra: ${monto_compra:,.2f}')
    print(f'Monto del descuento: ${monto_descuento:,.2f}')
    print(f'Monto final de la compra con descuento: ${monto_final:,.2f}')
else:
    print('\nNo obtuviste ningún tipo de descuento')
    print('Te invitamos a hacerte miembro de la tienda')
    print(f'Monto final de la compra: ${monto_compra:,.2f}')
```
