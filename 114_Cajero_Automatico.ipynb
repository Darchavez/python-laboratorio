# 🧪 114 — Aplicación Cajero Automático

**Aprendiendo:** ciclo `while` con una variable de control (`salir`), `while...else`, operadores de asignación (`+=` y `-=`), condicionales anidados y validación de montos.

## Enunciado
Crear una aplicación de cajero automático con estas operaciones:

- Consultar el saldo
- Retirar: se resta del saldo (solo si hay saldo suficiente)
- Depositar: se suma al saldo
- Salir

El saldo inicial es de $1,000.00.

## 1. Definir el saldo inicial

```python
print('*** Aplicación de Cajero Automático ***')

SALDO_INICIAL = 1000.00

saldo = SALDO_INICIAL
salir = False
```

## 2. Menú del cajero con un ciclo while

```python
while not salir:
    print('''\n--- Operaciones que puedes realizar ---
1. Consultar saldo
2. Retirar
3. Depositar
4. Salir''')
    opcion = int(input('Escoge una opción: '))

    if opcion == 1:  # Consultar saldo
        print(f'Tu saldo actual es: ${saldo:,.2f}')

    elif opcion == 2:  # Retirar
        retiro = float(input('Ingresa el monto a retirar: '))
        if retiro <= 0:
            print('El monto debe ser mayor que 0.')
        elif retiro <= saldo:
            saldo -= retiro  # saldo = saldo - retiro
            print(f'Retiro exitoso. Tu nuevo saldo es: ${saldo:,.2f}')
        else:
            print(f'Saldo insuficiente. Tu saldo actual es: ${saldo:,.2f}')

    elif opcion == 3:  # Depositar
        deposito = float(input('Ingresa el monto a depositar: '))
        if deposito <= 0:
            print('El monto debe ser mayor que 0.')
        else:
            saldo += deposito  # saldo = saldo + deposito
            print(f'Depósito exitoso. Tu nuevo saldo es: ${saldo:,.2f}')

    elif opcion == 4:  # Salir
        print('Saliendo del cajero automático. ¡Hasta pronto!')
        salir = True

    else:
        print('Opción inválida. Selecciona una opción del 1 al 4.')
else:
    print('Terminando la aplicación de Cajero Automático.')
```
