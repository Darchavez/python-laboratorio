# 🧪 151 — Gestión de Inventario

**Aprendiendo:** listas para guardar varios registros, diccionarios para el detalle de cada producto, recorrer una lista con `for` y agregar elementos con `append()`.

## Enunciado
Crear un programa para gestionar el inventario de un almacén.

- Utilizar una **lista** de Python para mantener un registro de los productos disponibles en el almacén.
- Para almacenar el detalle de cada producto, utilizar un **diccionario** con: id, nombre, precio y cantidad disponible en almacén.

## 1. Crear el inventario
Cada producto es un **diccionario** (con sus llaves `id`, `nombre`, `precio` y `cantidad`), y todos los productos se guardan dentro de una **lista**.

```python
print('*** Gestión de Inventario ***')

inventario = [
    {'id': 1, 'nombre': 'Camisa', 'precio': 25.99, 'cantidad': 50},
    {'id': 2, 'nombre': 'Pantalón', 'precio': 39.99, 'cantidad': 30},
    {'id': 3, 'nombre': 'Zapatos', 'precio': 49.99, 'cantidad': 20},
]
```

## 2. Mostrar el inventario

```python
print('\n--- Inventario del almacén ---')
for producto in inventario:
    print(f"Id: {producto['id']} | Nombre: {producto['nombre']} | "
          f"Precio: ${producto['precio']:,.2f} | Cantidad: {producto['cantidad']}")
```

## 3. Agregar un nuevo producto
Se piden los datos, se arma un diccionario nuevo y se agrega al final de la lista con `append()`.

```python
print('\n--- Agregar nuevo producto ---')
id_producto = int(input('Id: '))
nombre = input('Nombre: ')
precio = float(input('Precio: '))
cantidad = int(input('Cantidad: '))

nuevo_producto = {'id': id_producto, 'nombre': nombre, 'precio': precio, 'cantidad': cantidad}
inventario.append(nuevo_producto)
print('Producto agregado al inventario.')
```

## 4. Mostrar el inventario actualizado

```python
print('\n--- Inventario actualizado ---')
for producto in inventario:
    print(f"Id: {producto['id']} | Nombre: {producto['nombre']} | "
          f"Precio: ${producto['precio']:,.2f} | Cantidad: {producto['cantidad']}")

print(f'\nTotal de productos registrados: {len(inventario)}')
```
