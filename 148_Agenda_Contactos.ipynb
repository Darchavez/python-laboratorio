# 🧪 148 — Agenda de Contactos

**Aprendiendo:** diccionarios anidados (un diccionario dentro de otro), recorrerlos con `items()`, agregar llaves nuevas y buscar con `in` y `get()`.

## Enunciado
Crear una agenda de contactos utilizando un diccionario de Python con la siguiente estructura:

```
agenda {
    nombre {
        telefono
        email
        direccion
    }
}
```

## 1. Crear la agenda
La **llave** de la agenda es el nombre del contacto, y su **valor** es otro diccionario con el teléfono, el email y la dirección.

````python
print('*** Agenda de Contactos ***')

agenda = {
    'Ana': {'telefono': '809-555-1234', 'email': 'ana@mail.com', 'direccion': 'Santo Domingo'},
    'Luis': {'telefono': '829-555-5678', 'email': 'luis@mail.com', 'direccion': 'Santiago'},
}
````

## 2. Mostrar la agenda
`items()` devuelve cada nombre junto con su diccionario de datos.

````python
print('\n--- Contactos ---')
for nombre, datos in agenda.items():
    print(f"{nombre} | Tel: {datos['telefono']} | Email: {datos['email']} | Dirección: {datos['direccion']}")
````

## 3. Agregar un contacto
Para agregar, basta con asignar un valor a una llave nueva: `agenda[nombre] = {...}`.

````python
print('\n--- Agregar contacto ---')
nombre = input('Nombre: ').strip().title()
telefono = input('Teléfono: ').strip()
email = input('Email: ').strip().lower()
direccion = input('Dirección: ').strip()

agenda[nombre] = {'telefono': telefono, 'email': email, 'direccion': direccion}
print(f'{nombre} fue agregado a la agenda.')
````

## 4. Buscar un contacto
`get()` devuelve los datos del contacto, o `None` si no existe, sin dar error.

````python
nombre_buscar = input('\nNombre a buscar: ').strip().title()
contacto = agenda.get(nombre_buscar)

if contacto:
    print(f"\n{nombre_buscar} | Tel: {contacto['telefono']} | Email: {contacto['email']} | "
          f"Dirección: {contacto['direccion']}")
else:
    print(f'{nombre_buscar} no está en la agenda.')

print(f'\nTotal de contactos: {len(agenda)}')
````
