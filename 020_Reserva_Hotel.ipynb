# 020 — Sistema de Reserva de Hoteles

**Aprendiendo:** variables de distintos tipos, imprimirlas y cambiar su valor.

## Enunciado
Crea un sistema de reserva de hoteles que contenga la siguiente información de una reserva:
- Nombre del cliente
- Días de estancia
- Tarifa diaria
- Si el cuarto tiene vista al mar

Imprime el valor de cada variable. Luego haz cambios y vuelve a imprimir.

## 1. Crear las variables

```python
print('*** Sistema de Reserva de Hoteles ***')
# Definimos las variables
nombre_cliente = 'Laura Martínez'
dias_estancia = 5
tarifa_diaria = 1200.00
vista_al_mar = True
```

## 2. Mostrar la reserva

```python
# Mostrar el detalle de la reserva
print('Cliente:', nombre_cliente)
print('Días de estancia:', dias_estancia)
print(f'Tarifa diaria: ${tarifa_diaria:.2f}')
print('¿Habitación con vista al mar?', vista_al_mar)
```

## 3. Modificar valores

```python
# Realizamos algunas modificaciones
dias_estancia = 7
tarifa_diaria = 1000.00
vista_al_mar = False
```

## 4. Mostrar la reserva modificada

```python
# Mostrar el detalle de la reserva
print()
print('Cliente:', nombre_cliente)
print('Días de estancia:', dias_estancia)
print(f'Tarifa diaria: ${tarifa_diaria:.2f}')
print('¿Habitación con vista al mar?', vista_al_mar)
```
