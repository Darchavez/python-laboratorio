# 053 — ID Unico generador

**Aprendiendo:** importar funciones (`from random import randint`), generar números aleatorios con `randint()`, limpiar texto con `strip()` y `upper()`, cortar cadenas con `[0:2]` y unir todo con f-strings.

## Enunciado
Se solicita crear un sistema para generar un ID único para cada persona. El sistema debe solicitar al usuario:

- Nombre
- Apellido
- Año de nacimiento (YYYY)

El sistema deberá generar un valor aleatorio de 4 dígitos con la función randint.

## 1. Crear sistema para generar el valor de id unico

```python
from random import randint

print('*** Sistema Generador de ID Único ***')

nombre = input('Cual es tu nombre? ')
apellido = input('Cual es tu apellido? ')
anio_nacimiento = input('Cual es tu año de nacimiento (YYYY)? ') # Y - year
```

## 2. Normalizar los valores

```python
# Normalizar los valores
nombre_2 = nombre.strip().upper()[0:2]
apellido_2 = apellido.strip().upper()[0:2]
anio_nacimiento_2 = anio_nacimiento.strip()[2:] # Tambien puede ser [2:4]
```

## 3. Generar el valor aleatorio

```python
# Generar el valor aleatorio
aleatorio = randint(1000, 9999)
```

## 4. Generamos el valor de id unico

```python
# Generamos el valor de id unico
id_unico = f'{nombre_2}{apellido_2}{anio_nacimiento_2}{aleatorio}'

print(f'''\nHola {nombre.capitalize()},
    Tu nuevo número de identificación (ID) generado por el sistema es:
    {id_unico}
    Felicidades!''')
```
