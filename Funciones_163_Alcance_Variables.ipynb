# 🧪 163 — Alcance de Variables

**Aprendiendo:** la diferencia entre variables **globales** (existen en todo el programa) y **locales** (solo existen dentro de una función), y la palabra clave `global` para modificar una variable global desde dentro de una función.

## Enunciado
Crear un programa con un contador global y una función que tenga su propio contador local.

Cada vez que se llame la función, se deben incrementar ambos contadores e imprimir sus valores. Llamar la función varias veces y, al final, imprimir el valor de la variable global para comparar cómo se comporta cada una.

## 1. Definir la variable global

```python
print('*** Alcance de Variables ***')

# Variable global
contador_global = 0
```

## 2. Definir la función
`global contador_global` le avisa a Python que vamos a usar la variable de **afuera**. Sin esa línea, Python intentaría crear una variable local nueva con el mismo nombre y daría error al sumarle 1.

```python
def incrementar_contador():
    # Usar la variable global
    global contador_global
    # Declaramos una variable local
    contador_local = 0
    # Incrementamos la variable global
    contador_global += 1
    # Incrementamos la variable local
    contador_local += 1
    # Imprimimos ambos contadores
    print(f'Contador local: {contador_local}')
    print(f'Contador global: {contador_global}\n')
```

## 3. Llamar varias veces la función

```python
incrementar_contador()
incrementar_contador()
incrementar_contador()
```

## 4. Ver el valor final de la variable global

```python
# Terminando el programa
print(f'Valor variable global: {contador_global}')
```
