# 🧪 105 — Sistema de Autenticación (4 casos)

**Aprendiendo:** constantes, operadores de comparación (`==`), operadores lógicos (`and`, `not`) y condicionales (`if/elif/else`) para manejar varios casos.

## Enunciado
Crear un sistema para validar los valores de usuario y password proporcionados.

Se deben definir dos constantes con los valores válidos de usuario y password, y el sistema debe comparar los valores válidos contra los valores proporcionados.

Se deben considerar 4 casos:

| Caso | Mensaje |
|---|---|
| 1. Usuario y password válidos | Bienvenido al Sistema |
| 2. Usuario inválido | Usuario inválido |
| 3. Password inválido | Password inválido |
| 4. Usuario y password inválidos | Usuario y password inválidos |

## 1. Definir las constantes

```python
print('*** Sistema de Autenticación ***')

USUARIO_VALIDO = 'admin'
PASSWORD_VALIDO = '123'
```

## 2. Pedir los datos al usuario

```python
usuario_ingresado = input('¿Cuál es tu usuario? ').strip()
password_ingresado = input('¿Cuál es tu password? ').strip()
```

## 3. Validar cada dato por separado

```python
usuario_correcto = usuario_ingresado == USUARIO_VALIDO
password_correcto = password_ingresado == PASSWORD_VALIDO
```

## 4. Revisar los 4 casos

```python
if usuario_correcto and password_correcto:
    print('Bienvenido al Sistema')
elif not usuario_correcto and not password_correcto:
    print('Usuario y password inválidos')
elif not usuario_correcto:
    print('Usuario inválido')
else:
    print('Password inválido')
```
