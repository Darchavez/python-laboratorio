# 🧪 206 — Clase Aritmética con Propiedades (get y set)

**Aprendiendo:** **encapsulamiento** con atributos "protegidos" (`_operando1`), **propiedades** con `@property` (get, para leer) y `@setter` (set, para modificar), y **valores por defecto** en el constructor (`=None`) para crear objetos con todos, algunos o ningún dato.

## Enunciado
Mejorar la clase `Aritmetica` para que:
- Sus operandos se guarden como atributos protegidos.
- Se puedan **leer** y **modificar** desde afuera usando propiedades.
- Se puedan crear objetos de distintas formas:

| Objeto | Cómo se crea |
|---|---|
| 1 | Con los dos operandos, y luego se modifican |
| 2 | Con los dos operandos |
| 3 | Con un solo operando; el otro se asigna después |
| 4 | Sin operandos; los dos se asignan después |
| 5 | Con los operandos por nombre, en otro orden |

## 1. Definir la clase
- **`=None` en el constructor:** si no se envía un operando, empieza vacío (`None`) en vez de dar error.
- **El guion bajo `_operando1`:** es un aviso entre programadores: *"no toques esto directamente desde afuera"*.
- **`@property`** (get): permite **leer** el valor con `objeto.operando1`.
- **`@operando1.setter`** (set): permite **cambiar** el valor con `objeto.operando1 = 9`.

```python
class Aritmetica:

    def __init__(self, operando1=None, operando2=None):
        self._operando1 = operando1
        self._operando2 = operando2

    def sumar(self):
        resultado = self._operando1 + self._operando2
        print(f'Resultado suma: {resultado}')

    def restar(self):
        resultado = self._operando1 - self._operando2
        print(f'Resultado resta: {resultado}')

    def multiplicar(self):
        resultado = self._operando1 * self._operando2
        print(f'Resultado multiplicación: {resultado}')

    def dividir(self):
        resultado = self._operando1 / self._operando2
        print(f'Resultado división: {resultado}')

    # Propiedad operando1 (get y set)
    @property
    def operando1(self):
        return self._operando1

    @operando1.setter
    def operando1(self, operando1):
        self._operando1 = operando1

    # Propiedad operando2 (get y set)
    @property
    def operando2(self):
        return self._operando2

    @operando2.setter
    def operando2(self, operando2):
        self._operando2 = operando2
```

## 2. Probar los objetos
Fíjate que desde afuera siempre se usa `operando1` (sin guion bajo). Por dentro, la propiedad es la que lee o cambia `_operando1`.

```python
if __name__ == '__main__':
    print('*** Ejemplo Clase Aritmética ***')

    # Primer objeto: con los dos operandos, y luego se modifican
    print('\n--- Primer objeto ---')
    aritmetica1 = Aritmetica(5, 7)
    print(f'Valor operando1 del objeto aritmetica1: {aritmetica1.operando1}')
    print(f'Valor operando2 del objeto aritmetica1: {aritmetica1.operando2}')
    aritmetica1.sumar()
    aritmetica1.restar()
    aritmetica1.operando1 = 9
    aritmetica1.operando2 = 15
    print(f'Nuevo valor operando1 del objeto aritmetica1: {aritmetica1.operando1}')
    print(f'Nuevo valor operando2 del objeto aritmetica1: {aritmetica1.operando2}')
    aritmetica1.sumar()
    aritmetica1.restar()

    # Segundo objeto: con los dos operandos
    print('\n--- Segundo objeto ---')
    aritmetica2 = Aritmetica(12, 16)
    print(f'Valor operando1 del objeto aritmetica2: {aritmetica2.operando1}')
    print(f'Valor operando2 del objeto aritmetica2: {aritmetica2.operando2}')
    aritmetica2.sumar()
    aritmetica2.restar()

    # Tercer objeto: con un solo operando
    print('\n--- Tercer objeto ---')
    aritmetica3 = Aritmetica(7)
    aritmetica3.operando2 = 9
    aritmetica3.sumar()

    # Cuarto objeto: sin operandos
    print('\n--- Cuarto objeto ---')
    aritmetica4 = Aritmetica()
    aritmetica4.operando1 = 2
    aritmetica4.operando2 = 8
    aritmetica4.sumar()

    # Quinto objeto: operandos por nombre, en otro orden
    print('\n--- Quinto objeto ---')
    aritmetica5 = Aritmetica(operando2=4, operando1=3)
    aritmetica5.sumar()
```
