# 🧪 120 — Juego de Adivinanzas

**Aprendiendo:** generar números aleatorios con `randint()`, ciclo `while` con una variable de control (`adivinado`), condicionales anidados y un contador que se incrementa con `+=`.

## Enunciado
Crear un juego donde el jugador debe adivinar un número secreto.

- El número secreto se crea con la función `randint`, con un valor aleatorio entre 1 y 50.
- Usar un ciclo `while` que se repita hasta que el jugador adivine correctamente.
- Por cada intento fallido, incrementar una variable que lleve el conteo de intentos.

## 1. Definir las constantes y generar el número secreto

```python
from random import randint

print('*** Juego de Adivinanzas ***')

NUMERO_MINIMO = 1
NUMERO_MAXIMO = 50

numero_secreto = randint(NUMERO_MINIMO, NUMERO_MAXIMO)
intentos_fallidos = 0
adivinado = False
```

## 2. Ciclo del juego

```python
while not adivinado:
    numero_jugador = int(input(f'Adivina el número ({NUMERO_MINIMO} al {NUMERO_MAXIMO}): '))

    if numero_jugador == numero_secreto:
        adivinado = True
    else:
        intentos_fallidos += 1  # intentos_fallidos = intentos_fallidos + 1
        if numero_jugador < numero_secreto:
            print('Incorrecto. El número secreto es MAYOR.')
        else:
            print('Incorrecto. El número secreto es MENOR.')

print(f'\n¡Felicidades! Adivinaste el número secreto: {numero_secreto}')
print(f'Intentos fallidos: {intentos_fallidos}')
```
