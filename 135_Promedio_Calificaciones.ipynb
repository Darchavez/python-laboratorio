# 🧪 135 — Promedio de Calificaciones

**Aprendiendo:** ciclo `for` con `range()`, un acumulador que suma con `+=`, conversión de datos con `int()` y `float()` y cálculo de un promedio.

## Enunciado
Crear un programa para calcular el promedio de calificaciones.

- El programa debe solicitar el número de calificaciones a utilizar para obtener el promedio.
- Posteriormente, debe solicitar cada calificación al usuario.
- Después, realizar la suma de todas las calificaciones y, finalmente, imprimir el promedio.

## 1. Pedir el número de calificaciones

```python
print('*** Promedio de Calificaciones ***')

numero_calificaciones = int(input('¿Cuántas calificaciones vas a ingresar? '))
suma_calificaciones = 0
```

## 2. Pedir cada calificación y sumarlas

```python
for i in range(1, numero_calificaciones + 1):
    calificacion = float(input(f'Calificación {i}: '))
    suma_calificaciones += calificacion  # suma_calificaciones = suma_calificaciones + calificacion
```

## 3. Calcular e imprimir el promedio

```python
if numero_calificaciones > 0:
    promedio = suma_calificaciones / numero_calificaciones
    print(f'\nSuma de las calificaciones: {suma_calificaciones:.2f}')
    print(f'Promedio: {promedio:.2f}')
else:
    print('Debes ingresar al menos una calificación.')
```
